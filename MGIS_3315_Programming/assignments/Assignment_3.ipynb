```python
#created a class to be the blueprint to create objects.
#added two methods to calculate area and perimeter. Methods are functions that inside a class. 
#__iit__ is a constructor for instance variables.
class Rectangle():
    def __init__(self, length, width):
        self.length=length
        self.width=width
    def calculateAreas(self):
        return self.length * self.width
    def calculatePerimeter(self):
        return (self.length*2)+(self.width*2)
#square is listed as an object and it calls on the Rectangle class.
#the "(10,10) refer to the variables set from __init__.
square = Rectangle(10,10)
#added "+" symbols to tell python I want to add my functions together with my strings. 
print("The area for our square is " + str(square.calculateAreas()) + "cm and the perimeter is " + str(square.calculatePerimeter()) + "cm.")
```

```python
#Student class is made the parent class. 
class Student():
    university="ABC university"
    def __init__(self, name, sid):
        self.name=name
        self.sid=sid
    def displayName(self):
        return self.name
#Grade class inherits the Student class and super().__init__ referes to the variables set in the Student class.
#Grade class has a new variable called courseGrade.
#displayGrade function refers to the courseGrade variable in the object and gives a letter. 
class Grade(Student):
    def __init__(self,name,sid,courseGrade):
        super().__init__(name,sid)
        self.courseGrade=courseGrade
    def displayGrade(self):
        if self.courseGrade >= 89.45:
            letter = "A"
        elif self.courseGrade <=89.44 and self.courseGrade >= 79.45:
            letter = "B"
        elif self.courseGrade <=79.44 and self.courseGrade >= 69.45:
            letter = "C"
        elif self.courseGrade <=69.44 and self.courseGrade >= 59.45:
            letter = "D"
        elif self.courseGrade <=59.44:
            letter = "F"
        return letter
#object that the above code refers to. Each variable totals to three, which matches the Grade class. 
student = Grade("Tom","s1",75.3)
#printing functions and strings to form a sentence. 
print("The letter grade from " +student.displayName()+ "'s course at " +student.university+ " is a " +student.displayGrade()+".")
```

```python

#set Employee class with two variables. 
#added __lt__ for less than. 
#added __gt__ for greater than.
# lt and gt methods are used to compare the salary variable of the two objects. 
class Employee():
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
    def __lt__(self,obj2):
        return self.salary < obj2.salary
    def __gt__(self,obj2):
        return self.salary > obj2.salary

John = Employee("John",5000.50)
Maria = Employee("Maria",4090.50)
#Results from the objects when compared. 
print("Is Maria's salary less than John?", Maria<John)
print("Is John's salary greater than Maria?", John>Maria)
```
